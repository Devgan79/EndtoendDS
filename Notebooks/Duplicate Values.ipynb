<h1>📘 Handling Duplicate Values in Pandas</h1>
<p>A complete theory + examples + coding reference</p>


<!-- SECTION 1 -->
<div class="section">
<h2>1. What Are Duplicate Values?</h2>

<p>
Duplicate values occur when two or more rows in a dataset contain
the same information.
</p>

<p>
Duplicates usually happen because of:
</p>

<ul>
    <li>Repeated data entry</li>
    <li>System/database errors</li>
    <li>Merging multiple datasets</li>
    <li>Data scraping issues</li>
    <li>Network or logging failures</li>
</ul>

<p>
If duplicates are not handled properly, they can:
</p>

<ul>
    <li>Bias machine learning models</li>
    <li>Produce wrong statistics</li>
    <li>Increase storage size</li>
    <li>Reduce data quality</li>
</ul>

<div class="note">
Always analyze duplicates before removing them.
</div>
</div>


<!-- SECTION 2 -->
<div class="section">
<h2>2. Types of Duplicates</h2>

<h3>1️⃣ Full Duplicates</h3>
<p>
All columns are exactly the same.
</p>

<h3>2️⃣ Partial Duplicates</h3>
<p>
Some columns are the same, some are different.
</p>

<h3>3️⃣ Key-Based Duplicates</h3>
<p>
Duplicates based on unique identifiers (ID, email, phone).
</p>

<h3>4️⃣ Time-Series Duplicates</h3>
<p>
Same timestamp appears more than once.
</p>

<div class="example">
Example: Two rows have same ID but different address → Partial duplicate.
</div>
</div>


<!-- SECTION 3 -->
<div class="section">
<h2>3. Detect Duplicate Rows</h2>

<p>
Before deleting, always find duplicates.
</p>

<code>
# Count duplicate rows
df.duplicated().sum()

# See True/False mask
df.duplicated()

# View duplicate rows
df[df.duplicated()]

# View all copies
df[df.duplicated(keep=False)]
</code>
</div>


<!-- SECTION 4 -->
<div class="section">
<h2>4. Check Duplicates in Selected Columns</h2>

<p>
Often, duplicates are meaningful only in some columns.
</p>

<code>
# Based on ID column
df.duplicated(subset=['id']).sum()

# Based on multiple columns
df.duplicated(subset=['name', 'dob']).sum()
</code>

<div class="example">
Email addresses are usually unique → use them to find duplicates.
</div>
</div>


<!-- SECTION 5 -->
<div class="section">
<h2>5. Inspect Duplicate Groups</h2>

<p>
Inspecting helps decide which row to keep.
</p>

<code>
df[df.duplicated(subset=['email'], keep=False)]
  .sort_values('email')
</code>
</div>


<!-- SECTION 6 -->
<div class="section">
<h2>6. Removing Duplicate Rows</h2>

<h3>Keep First (Default)</h3>
<code>
df.drop_duplicates(inplace=True)
</code>

<h3>Keep Last</h3>
<code>
df.drop_duplicates(keep='last', inplace=True)
</code>

<h3>Remove All Copies</h3>
<code>
df.drop_duplicates(keep=False, inplace=True)
</code>

<div class="warning">
Removing all copies may delete valid data.
</div>
</div>


<!-- SECTION 7 -->
<div class="section">
<h2>7. Remove Duplicates by Columns</h2>

<code>
# Keep one record per ID
df.drop_duplicates(subset=['id'], inplace=True)

# Multiple columns
df.drop_duplicates(subset=['name','phone'], inplace=True)
</code>
</div>


<!-- SECTION 8 -->
<div class="section">
<h2>8. Merging Partial Duplicates</h2>

<p>
Sometimes duplicate rows contain useful information.
</p>

<div class="example">
Row 1: Phone missing
Row 2: Address missing
→ Combine both into one record
</div>

<code>
# Basic merge
df = df.groupby('id').agg('first').reset_index()

# Smart merge (skip NaN)
df = df.groupby('id')
       .agg(lambda x: x.dropna().iloc[0])
       .reset_index()
</code>
</div>


<!-- SECTION 9 -->
<div class="section">
<h2>9. Time-Series Duplicate Handling</h2>

<p>
Duplicate timestamps cause analysis errors.
</p>

<code>
# Remove duplicate index values
df = df[~df.index.duplicated(keep='first')]

# Or using column
df = df.drop_duplicates(subset=['date'])
</code>
</div>


<!-- SECTION 10 -->
<div class="section">
<h2>10. Mark Duplicates Instead of Removing</h2>

<p>
Sometimes you want to analyze duplicates.
</p>

<code>
df['is_duplicate'] = df.duplicated().astype(int)
</code>
</div>


<!-- SECTION 11 -->
<div class="section">
<h2>11. Find Most Repeated Records</h2>

<code>
# Most frequent rows
df.value_counts().head()

# Group-wise count
df.groupby(df.columns.tolist())
  .size()
  .sort_values(ascending=False)
</code>
</div>


<!-- SECTION 12 -->
<div class="section">
<h2>12. Real-World Cleaning Pipeline</h2>

<p>
Recommended for projects.
</p>

<code>
# Backup data
df_backup = df.copy()

# Check duplicates
print("Duplicates:", df.duplicated().sum())

# Remove full duplicates
df.drop_duplicates(inplace=True)

# Remove key-based duplicates
df.drop_duplicates(subset=['id'], keep='first', inplace=True)

# Reset index
df.reset_index(drop=True, inplace=True)

# Verify
assert df.duplicated().sum() == 0
</code>
</div>


<!-- SECTION 13 -->
<div class="section">
<h2>13. When NOT to Remove Duplicates</h2>

<ul>
    <li>Transaction records</li>
    <li>Log files</li>
    <li>Clickstream data</li>
    <li>Sensor events</li>
</ul>

<div class="warning">
In these cases, duplicates represent real events.

<h1>📊 Duplicate Handling in Kitchen Appliance Time-Series Logs</h1>
<p>Theory + Examples + Pandas Code</p>


<!-- SECTION 1 -->
<div class="section">
<h2>1. What Are Kitchen Appliance Log Files?</h2>

<p>
Modern kitchen appliances generate continuous log data.
These logs are collected from sensors and controllers.
</p>

<p>
Examples of devices:
</p>

<ul>
    <li>Smart refrigerators</li>
    <li>Smart ovens</li>
    <li>Dishwashers</li>
    <li>Microwaves</li>
    <li>Coffee machines</li>
</ul>

<p>
Typical log structure:
</p>

<code>
timestamp | device_id | temperature | power | status | error_code
</code>

<div class="note">
Each row represents one device reading at a given time.
</div>
</div>


<!-- SECTION 2 -->
<div class="section">
<h2>2. Why Do Duplicates Appear in IoT Logs?</h2>

<p>
Duplicate records are very common in IoT systems.
</p>

<p>
Main reasons:
</p>

<ul>
    <li>Network re-transmission</li>
    <li>Power interruptions</li>
    <li>Device reboots</li>
    <li>Cloud sync failures</li>
    <li>Sensor buffering</li>
</ul>

<div class="example">
If Wi-Fi disconnects, the oven may resend old data.
</div>
</div>


<!-- SECTION 3 -->
<div class="section">
<h2>3. Types of Duplicates in Kitchen Logs</h2>

<h3>1️⃣ Exact Duplicates</h3>
<p>All columns are identical.</p>

<h3>2️⃣ Timestamp Duplicates</h3>
<p>Same time + same device appears multiple times.</p>

<h3>3️⃣ Partial Duplicates</h3>
<p>Same time, different sensor values.</p>

<h3>4️⃣ Burst Duplicates</h3>
<p>Same record repeated many times in seconds.</p>
</div>


<!-- SECTION 4 -->
<div class="section">
<h2>4. Example Log Dataset</h2>

<code>
timestamp           device_id   temp   power   status
------------------------------------------------------
2026-01-10 10:00:01  OVEN_01     180     ON      RUNNING
2026-01-10 10:00:01  OVEN_01     180     ON      RUNNING
2026-01-10 10:00:02  OVEN_01     181     ON      RUNNING
2026-01-10 10:00:02  OVEN_01     180     ON      RUNNING
2026-01-10 10:00:03  OVEN_01     182     ON      RUNNING
</code>

<div class="example">
Row 1 & 2 → Exact duplicate
Row 3 & 4 → Partial duplicate
</div>
</div>


<!-- SECTION 5 -->
<div class="section">
<h2>5. Prepare Log Data for Analysis</h2>

<code>
import pandas as pd

# Convert timestamp
df['timestamp'] = pd.to_datetime(df['timestamp'])

# Sort by time
df.sort_values('timestamp', inplace=True)

# Set index
df.set_index('timestamp', inplace=True)
</code>

<div class="note">
Sorting is mandatory before duplicate handling.
</div>
</div>


<!-- SECTION 6 -->
<div class="section">
<h2>6. Detect Exact Duplicates</h2>

<code>
# Count full duplicates
df.duplicated().sum()

# View them
df[df.duplicated()]
</code>
</div>


<!-- SECTION 7 -->
<div class="section">
<h2>7. Detect Timestamp + Device Duplicates</h2>

<p>
In IoT, this is the most important check.
</p>

<code>
# Check duplicates by time + device
df.duplicated(subset=['device_id'], keep=False)
</code>

<p>(Index is timestamp)</p>

<code>
df.reset_index()
  .duplicated(subset=['timestamp','device_id'])
  .sum()
</code>
</div>


<!-- SECTION 8 -->
<div class="section">
<h2>8. Removing Exact Duplicates (Safe Method)</h2>

<code>
# Keep first occurrence
df.drop_duplicates(inplace=True)
</code>

<div class="note">
Safe when values are truly identical.
</div>
</div>


<!-- SECTION 9 -->
<div class="section">
<h2>9. Handling Partial Duplicates (Aggregation)</h2>

<p>
When readings differ slightly, average them.
</p>

<code>
df = df.reset_index()

df = df.groupby(['timestamp','device_id']).agg({
    'temp': 'mean',
    'power': 'first',
    'status': 'last'
}).reset_index()

df.set_index('timestamp', inplace=True)
</code>

<div class="example">
Multiple readings at same second → merged into one.
</div>
</div>


<!-- SECTION 10 -->
<div class="section">
<h2>10. Handling Burst Duplicates</h2>

<p>
Burst duplicates happen when data repeats rapidly.
</p>

<code>
# Remove rows within 1 second window
df = df[~(
    (df['device_id'] == df['device_id'].shift()) &
    (df.index == df.index.shift())
)]
</code>
</div>


<!-- SECTION 11 -->
<div class="section">
<h2>11. Keep Latest Sensor Reading</h2>

<p>
Sometimes latest data is most accurate.
</p>

<code>
df = df.reset_index()

df = df.sort_values('timestamp')

df = df.drop_duplicates(
    subset=['timestamp','device_id'],
    keep='last'
)

df.set_index('timestamp', inplace=True)
</code>
</div>


<!-- SECTION 12 -->
<div class="section">
<h2>12. Mark Duplicate Events</h2>

<p>
Useful for debugging device behavior.
</p>

<code>
df['is_duplicate'] = df.reset_index()
    .duplicated(['timestamp','device_id'])
    .astype(int)
    .values
</code>
</div>


<!-- SECTION 13 -->
<div class="section">
<h2>13. Production-Ready Cleaning Pipeline</h2>

<code>
# Backup
raw_df = df.copy()

# Sort
df.sort_index(inplace=True)

# Remove exact duplicates
df.drop_duplicates(inplace=True)

# Aggregate partial duplicates
df = df.reset_index()

df = df.groupby(['timestamp','device_id']).agg({
    'temp':'mean',
    'power':'first',
    'status':'last'
}).reset_index()

# Set index
df.set_index('timestamp', inplace=True)

# Verify
assert df.duplicated().sum() == 0
</code>
</div>


<!-- SECTION 14 -->
<div class="section">
<h2>14. When NOT to Remove Duplicates</h2>

<ul>
    <li>Alarm logs</li>
    <li>Error events</li>
    <li>Safety shutdowns</li>
    <li>Critical failures</li>
</ul>

<div class="warning">
Removing these may hide dangerous events.
</div>
</div>

<div class="section">
<h2>15. Best Practices for IoT Logs</h2>

<ul>
    <li>Always store raw logs</li>
    <li>Clean in a separate copy</li>
    <li>Document rules</li>
    <li>Monitor duplicate rate</li>
    <li>Alert on high duplication</li>
</ul>

<div class="note">
Duplicate spikes often indicate hardware failure.
</div>
